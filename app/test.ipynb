- MindSpore 张量创建与基础算子实验

```python
# 1. 常量张量
t1 = ms.Tensor([[1.1,2.2],[3.3,4.4]], dtype=ms.float32)
t2 = ms.Tensor([1,2,3], dtype=ms.int32)
print(f't1:\n{t1}\n')
print(f't2:\n{t2}\n')
```

```python
# 2. 随机张量
rand_t = ops.UniformReal()((2,3)) # [0,1)均匀分布
randn_t = ops.StandardNormal()((2,3)) # 标准正态分布
print(f'rand_t:\n{rand_t}\n')
print(f'randn_t:\n{randn_t}\n')
```

```python
# 3. 序列、零、全一张量
seq_t = ops.Range()(1,20,1)
zero_t = ops.Zeros()((2,3), ms.float32)
ones_t = ops.Ones()((3,3), ms.float32)
print(f'seq_t:\n{seq_t}\n')
print(f'zero_t:\n{zero_t}\n')
print(f'ones_t:\n{ones_t}\n')
```

```python
# 4. 基础数学运算
# 生成标准正态分布（均值0，方差1）的张量，shape=(2,3)
x = ops.StandardNormal()((2,3))
# 逐元素取绝对值
abs_x = ops.Abs()(x)

a = ops.StandardNormal()((2,3))
b = ops.StandardNormal()((2,3))

# 逐元素相加 a + b
add_res = ops.Add()(a,b)

# 逐元素相乘 a * b（不是矩阵乘法！）
mul_res = ops.Mul()(a,b)

# 逐元素除法 a / 2
div_res = ops.Div()(a, 2)

# 数值裁剪：把a里面所有值限制在 [-0.2, 0.2]
clip_res = ops.clip_by_value(a, -0.2, 0.2)

# 元素平方 a ** 2
pow_res = ops.Pow()(a, ms.Tensor(2))

print(f'x:\n{x}\n')
print(f'a:\n{a}\n')
print(f'b:\n{b}\n')
print(f'abs_x:\n{abs_x}\n')
print(f'add_res:\n{add_res}\n')
print(f'mul_res:\n{mul_res}\n')
print(f'div_res:\n{div_res}\n')
print(f'clip_res:\n{clip_res}\n')
print(f'pow_res:\n{pow_res}\n')
```

```python
# 矩阵乘法、矩阵向量乘法
# 生成2行3列标准正态随机矩阵m1
m1 = ops.StandardNormal()((2,3))
# 生成3行2列标准正态随机矩阵m2
m2 = ops.StandardNormal()((3,2))

# 矩阵乘法：(2,3) × (3,2)
mat_mul = ops.MatMul()(m1,m2)

# 生成长度为3的一维向量vec
vec = ops.StandardNormal()((3,))

# 矩阵‑向量乘法 m1(2,3) * vec(3,)
mat_vec = ops.mv(m1, vec)

print(f'm1:\n{m1}\n')
print(f'm2:\n{m2}\n')
print(f'vec:\n{vec}\n')
print(f'mat_mul 矩阵乘法结果:\n{mat_mul}\n')
print(f'mat_vec 矩阵向量乘法结果:\n{mat_vec}\n')
```

- MindSpore 自动求导 GradOperation 示例

```python
import mindspore as ms
from mindspore import ops, Parameter
from mindspore.nn import Cell
# 定义计算网络
class Net(Cell):
    def __init__(self):
        super().__init__()
        self.x = Parameter(ms.Tensor([2.0,3.0]), name="x")
    def construct(self):
        y = self.x**2 + 4*self.x + 2
        return ops.ReduceSum()(y)
net = Net()
grad_op = ops.GradOperation(get_by_list=True)
grad_fn = grad_op(net, net.trainable_params())
grad = grad_fn()
print("梯度结果：", grad[0])
```

- 简易线性回归训练（Cell 网络 + 自动梯度）

```python
import mindspore as ms
from mindspore import ops, Parameter, ParameterTuple
from mindspore.nn import Cell, MSELoss

# ---------------------- 数据集准备 ----------------------
# y = 2*x 理想模型，最优 w=2.0, b=0.0
x_train = ms.Tensor([[1.], [2.], [3.]], ms.float32)
y_train = ms.Tensor([[2.], [4.], [6.]], ms.float32)

# ---------------------- 定义线性网络Cell ----------------------
class LinearNet(Cell):
    def __init__(self):
        super().__init__()
        # 待训练参数 权重w、偏置b
        self.w = Parameter(ms.Tensor(0.0, ms.float32), name="weight")
        self.b = Parameter(ms.Tensor(0.0, ms.float32), name="bias")

    def construct(self, x):
        """前向传播：y = w*x + b"""
        return self.w * x + self.b

# 实例化网络
net = LinearNet()
# MSE均方误差损失函数
loss_fn = MSELoss()

# 超参数
lr = 0.01
epochs = 200   # 增加迭代轮次，收敛更充分

# Assign算子：原地更新Parameter参数
assign = ops.Assign()
# 获取全部可训练参数，封装为ParameterTuple
params = ParameterTuple(net.trainable_params())

# 构建自动求导算子，对params参数列表求梯度
grad_op = ops.GradOperation(get_by_list=True)


# ---------------------- 训练循环 ----------------------
for epoch in range(epochs):
    # 前向函数：返回损失loss
    def forward_fn(x):
        y_pred = net(x)
        loss = loss_fn(y_pred, y_train)
        return loss

    # 生成梯度函数
    grad_fn = grad_op(forward_fn, params)
    # 执行前向+反向，得到梯度
    grads = grad_fn(x_train)
    dw, db = grads

    # 梯度下降原地更新参数 w = w - lr*dw ; b = b - lr*db
    assign(net.w, net.w - lr * dw)
    assign(net.b, net.b - lr * db)

    # 每10个epoch打印训练信息
    if epoch % 10 == 0:
        current_loss = forward_fn(x_train)
        print(f"epoch {epoch:3d} | loss = {current_loss.asnumpy():.4f} | "
              f"w = {net.w.asnumpy():.4f} | b = {net.b.asnumpy():.4f}")


# ---------------------- 训练结束输出最终参数 ----------------------
print("\n==========训练完成，最终参数==========")
final_w = net.w.asnumpy()
final_b = net.b.asnumpy()
print(f"训练得到权重 w = {final_w:.4f}")
print(f"训练得到偏置 b = {final_b:.4f}")
print(f"理论最优值：w=2.0000 , b=0.0000")
```

- 两层简易神经网络前向反向传播

```python
import mindspore as ms
from mindspore import ops, Parameter, ParameterTuple

# 超参与维度设置
batch_n = 100
input_dim = 1000
hidden_dim = 100
output_dim = 10

# 生成随机输入与标签
x = ops.StandardNormal()((batch_n, input_dim))
y = ops.StandardNormal()((batch_n, output_dim))

# 权重，Parameter标记为可训练参数
w1 = Parameter(ops.StandardNormal()((input_dim, hidden_dim)), name="w1")
w2 = Parameter(ops.StandardNormal()((hidden_dim, output_dim)), name="w2")

lr = 1e-6
epoch_n = 20

# Assign算子用于原地更新Parameter
assign = ops.Assign()
params = ParameterTuple([w1, w2])
grad_op = ops.GradOperation(get_by_list=True)

# 前向函数放到循环外，不每轮重复定义
def forward():
    # 第一层：矩阵乘法 + ReLU激活
    h1 = ops.MatMul()(x, w1)
    h1 = ops.Maximum()(h1, ms.Tensor(0))
    # 第二层矩阵乘法
    y_pred = ops.MatMul()(h1, w2)
    # 平方求和损失
    loss = ops.ReduceSum()(ops.Square()(y_pred - y))
    return loss

# 构造求导函数
grad_fn = grad_op(forward, params)

# 训练循环
for epoch in range(epoch_n):
    loss = forward()
    grad_w1, grad_w2 = grad_fn()

    # 原地更新权重，保留Parameter属性
    assign(w1, w1 - lr * grad_w1)
    assign(w2, w2 - lr * grad_w2)

    print(f"Epoch:{epoch:2d}, Loss:{loss.asnumpy():.4f}")

# 训练结束输出信息
print("\n=====训练完成=====")
print(f"w1 张量形状: {w1.shape}")
print(f"w2 张量形状: {w2.shape}")
```
